I considered Customer Churn Prediction as a lower-risk alternative — it's closer to what I've already built, so it would be faster to execute with less new material to learn. I chose Fraud Detection instead because the imbalance-handling and explainability skills it requires are the more distinctive, portfolio-worthy addition, and they connect naturally to the False Positive/False Negative reasoning I already practiced on the cardiac monitoring project — the same idea of an asymmetric cost between error types, now applied with the tools (SMOTE, SHAP) to handle it properly when classes are imbalanced.


## 📝 3. Problem Statement

Fraudulent transactions are rare compared to legitimate ones, but each missed case can carry real financial cost, while too many false alarms erode trust in the system and waste review effort. The goal of this project is to build a machine learning model that flags transactions likely to be fraudulent, learned from historical transaction data, while explicitly accounting for the fact that fraud cases are a small minority of the dataset — a genuinely imbalanced classification problem, not just a standard binary classification task.

**Scope — in:** data cleaning and EDA on a public/synthetic transaction dataset; a baseline classifier plus at least one comparison model; imbalance-handling (e.g. SMOTE) with precision-recall-based evaluation; SHAP-based model explainability; deployment at a public URL.

**Scope — out:** real-time transaction processing or production-grade infrastructure; identifiable, non-public customer data; deep learning approaches, unless the selected dataset specifically requires them.


## 🏁 4. Definition of Done

Per the program's professional baseline, whatever the project type, this capstone is complete when it delivers:

1. A clean, documented Jupyter Notebook covering the full pipeline: EDA → preprocessing → modeling → evaluation
2. A trained model with reported, cross-validated metrics — precision, recall, F1, and ROC-AUC/PR-AUC, since accuracy alone is misleading under class imbalance
3. A working deployment at a public URL (Streamlit or FastAPI)
4. A GitHub repository with a clean README, `requirements.txt`, and saved model artifacts
5. A short technical write-up explaining the approach, results, and limitations

Restating this now, before any implementation, is deliberate — it's the same discipline I used starting Week 4: define what "done" and "good" mean before writing code, so effort stays aimed at the actual target across all four sprints instead of drifting.


### 📌 Interpretation

The Definition of Done treats accuracy explicitly as insufficient — this is consistent with what I already learned evaluating the cardiac monitoring model: a headline accuracy number can look fine while hiding a model that is actually bad at catching the minority class. For fraud specifically, where the positive class (fraud) is rare, this matters even more than it did there.

Imbalance handling (SMOTE) and SHAP explainability belong to later sprints once a baseline exists to compare against; pulling them into Sprint 1 would blur what "baseline" actually means and make later improvement harder to measure honestly.
