# Probability Basics

Probability measures the likelihood that an event will occur.

A probability always lies between **0** and **1**.

- **0** means the event is impossible.
- **1** means the event is certain.

The basic probability formula is:

$$
P(A)=\frac{\text{Number of Favorable Outcomes}}{\text{Total Number of Possible Outcomes}}
$$

This formula assumes that every outcome is equally likely.

## Example 1 — Rolling a Fair Die

Suppose we roll a fair six-sided die.

Question:

What is the probability of rolling a **4**?

There is:

- One favorable outcome.
- Six possible outcomes.

Therefore,

$$
P(4)=\frac{1}{6}
$$

which is approximately **0.167**.

```python
favorable = 1
total = 6

probability = favorable / total

print(f"Probability of rolling a 4: {probability:.3f}")
```

### Interpretation

The probability of rolling a 4 is approximately **0.167**, meaning that if we roll a fair die many times, we expect about one out of every six rolls to result in a 4.

This does not guarantee that every six rolls will contain exactly one 4. Instead, probability describes the expected long-term behavior of random events.

# Understanding the Probability Scale

The probability of an event can take any value between **0** and **1**.

| Probability | Meaning |
|-------------|---------|
| 0 | Impossible event |
| 0.25 | Unlikely event |
| 0.50 | Equally likely to happen or not happen |
| 0.75 | Likely event |
| 1 | Certain event |

Probability measures uncertainty rather than guaranteeing an outcome.

# Core Probability Rules

Probability follows a set of mathematical rules that help us calculate the likelihood of events.

The three fundamental probability rules are:

- Complement Rule
- Addition Rule
- Multiplication Rule

These rules are widely used in statistics, data analysis, and machine learning for reasoning under uncertainty.

## 1. Complement Rule

The complement rule calculates the probability that an event **does not occur**.

If **A** is an event, then its complement is written as **A'** or **Not A**.

### Formula

$$
P(\text{Not A}) = 1 - P(A)
$$

This rule is useful because sometimes calculating the probability of the opposite event is easier than calculating the event directly.

### Example

Suppose the probability that it rains today is:

$$
P(\text{Rain}) = 0.30
$$

What is the probability that it **does not rain**?

$$
P(\text{No Rain}) = 1 - 0.30 = 0.70
$$

```python
p_rain = 0.30

p_no_rain = 1 - p_rain

print(f"Probability of Rain: {p_rain}")
print(f"Probability of No Rain: {p_no_rain}")
```

### Interpretation

If there is a 30% chance of rain, then there must be a 70% chance that it will not rain.

The complement rule ensures that the probabilities of an event and its opposite always add up to **1**.

### Connection to Machine Learning

In binary classification problems, a model predicts the probability of one class.

For example:

- P(Diabetes) = 0.82

Then:

- P(No Diabetes) = 0.18

These probabilities always sum to **1**.

# Addition Rule

The addition rule calculates the probability that **at least one of two events occurs**.

When two events can happen together, we subtract the overlap to avoid counting it twice.

### Formula

$$
P(A \cup B)=P(A)+P(B)-P(A\cap B)
$$

### Example

Suppose:

- P(Student studies AI) = 0.60
- P(Student studies Cybersecurity) = 0.40
- P(Student studies both) = 0.20

What is the probability that a student studies **AI or Cybersecurity**?

```python
p_ai = 0.60
p_cyber = 0.40
p_both = 0.20

p_union = p_ai + p_cyber - p_both

print(f"P(AI or Cybersecurity): {p_union}")
```

### Interpretation

The overlap between the two events is subtracted because students enrolled in both subjects would otherwise be counted twice.

The addition rule ensures that every observation is counted only once.

### Connection to Machine Learning

The addition rule is useful when calculating the probability that an observation belongs to one of several categories or satisfies multiple conditions.

It is frequently used during probabilistic reasoning and model evaluation.

# Multiplication Rule

The multiplication rule calculates the probability that **two independent events occur together**.

Two events are independent if the occurrence of one event does not affect the probability of the other.

### Formula

$$
P(A\cap B)=P(A)\times P(B)
$$

### Example

Suppose we toss a fair coin and roll a fair die.

What is the probability of getting:

- Heads
- A six

Both events are independent.

```python
p_heads = 0.5
p_six = 1/6

p_both = p_heads * p_six

print(f"Probability of Heads and Six: {p_both:.3f}")
```

### Interpretation

The probability of both events occurring is obtained by multiplying their individual probabilities because the two events do not influence each other.

As more independent events are combined, the overall probability generally becomes smaller.

### Connection to Machine Learning

The multiplication rule is particularly important in the **Naive Bayes** classifier.

Naive Bayes assumes that features are conditionally independent, allowing the model to multiply probabilities when estimating the likelihood of each class.

# Summary of Probability Rules

| Rule | Formula | Purpose |
|------|---------|---------|
| Complement | \(P(\text{Not A}) = 1 - P(A)\) | Probability that an event does not occur |
| Addition | \(P(A \cup B)=P(A)+P(B)-P(A\cap B)\) | Probability that at least one event occurs |
| Multiplication | \(P(A\cap B)=P(A)\times P(B)\) | Probability that independent events occur together |

# Conditional Probability

Conditional probability measures the probability of an event occurring **given that another event has already occurred**.

Instead of asking:

> "What is the probability of event A?"

we ask:

> "What is the probability of event A, given that event B has already happened?"

Conditional probability is widely used in machine learning because predictions are usually made based on available information (features).

For example:

- Given that a patient has a high glucose level, what is the probability that they have diabetes?
- Given that an email contains the word "Free", what is the probability that it is spam?

## Formula

The conditional probability of **A given B** is calculated as:

$$
P(A|B)=\frac{P(A \cap B)}{P(B)}
$$

Where:

- \(P(A|B)\) = Probability of A given B
- \(P(A \cap B)\) = Probability that both A and B occur
- \(P(B)\) = Probability of event B

## Example

Suppose a university has:

- 100 students.
- 40 students study Artificial Intelligence.
- 25 students study Artificial Intelligence **and** know Python.

Question:

What is the probability that a student knows Python **given that** they study Artificial Intelligence?

```python
students_ai = 40
students_ai_python = 25

conditional_probability = students_ai_python / students_ai

print(f"P(Python | AI) = {conditional_probability:.2f}")
```

### Interpretation

The probability is **0.625**, meaning that **62.5%** of students studying Artificial Intelligence also know Python.

Notice that we are **not** considering all students in the university.

Instead, we only focus on students who study Artificial Intelligence.

This is why conditional probability changes depending on the available information.

### Connection to Machine Learning

Almost every classification model estimates a conditional probability.

Examples include:

- Probability of diabetes **given** medical measurements.
- Probability of spam **given** email contents.
- Probability of customer churn **given** customer behavior.

Machine learning models rarely make predictions without conditions.

Instead, predictions are based on the observed features of each sample.

## Why Do We Need Bayes' Theorem?

Sometimes we know:

\[
P(B|A)
\]

but we actually need:

\[
P(A|B)
\]

Bayes' Theorem allows us to reverse conditional probabilities and update our beliefs after observing new evidence.

This idea forms the foundation of the Naive Bayes classifier, one of the most popular probabilistic machine learning algorithms.

# Bayes' Theorem

Bayes' Theorem is a mathematical formula used to update probabilities when new evidence becomes available.

Instead of asking:

> "What is the probability of observing this evidence?"

Bayes asks:

> "Given this evidence, what is the probability that my hypothesis is true?"

This concept is fundamental in statistics, artificial intelligence, and machine learning.

## Formula

$$
P(A|B)=
\frac{P(B|A)\times P(A)}
{P(B)}
$$

Where:

- **P(A)** → Prior Probability
- **P(B|A)** → Likelihood
- **P(B)** → Evidence
- **P(A|B)** → Posterior Probability

## Understanding Bayes' Components

### Prior

The probability we believe **before** seeing any evidence.

---

### Likelihood

How likely the evidence is if our hypothesis is true.

---

### Evidence

The overall probability of observing the evidence.

---

### Posterior

The updated probability after considering the evidence.

## Example

Suppose:

- 1% of people have a disease.
- A medical test correctly detects the disease 95% of the time.
- The probability of getting a positive test result is 5%.

What is the probability that a person actually has the disease after testing positive?

```python
prior = 0.01

likelihood = 0.95

evidence = 0.05

posterior = (likelihood * prior) / evidence

print(f"Posterior Probability = {posterior:.3f}")
```

### Interpretation

Although the test is highly accurate, the disease is rare.

Bayes' Theorem combines the prior probability with the observed evidence to produce an updated probability.

This updated probability is called the **posterior probability**.

### Bayes in Machine Learning

Bayes' Theorem is the mathematical foundation of the **Naive Bayes** classifier.

Naive Bayes assumes that features are conditionally independent and uses Bayes' Theorem to estimate the probability of each class.

Despite its simplicity, Naive Bayes performs remarkably well in applications such as:

- Spam Detection
- Sentiment Analysis
- Text Classification
- Medical Diagnosis

# Probability Distributions

A probability distribution describes how the values of a random variable are expected to occur.

Instead of asking whether an event happens or not, a probability distribution tells us **how likely different outcomes are**.

Understanding data distributions is an essential part of Exploratory Data Analysis (EDA) because many statistical methods and machine learning algorithms make assumptions about the underlying distribution of the data.

In this section, we will explore three of the most common probability distributions:

- Normal Distribution
- Binomial Distribution
- Uniform Distribution

# Normal Distribution (Gaussian Distribution)

The normal distribution is the most common probability distribution in statistics.

It has a symmetric bell-shaped curve where:

- Most observations are concentrated around the mean.
- Values become less frequent as they move away from the center.
- The mean, median, and mode are equal.

Many real-world measurements approximately follow a normal distribution, including:

- Human height
- Exam scores
- Measurement errors
- Blood pressure

```python
# Generate 10,000 samples from a normal distribution
normal_data = np.random.normal(loc=0, scale=1, size=10000)
```

```python
plt.figure(figsize=(8,5))

plt.hist(normal_data,
         bins=30,
         edgecolor="black")

plt.title("Normal Distribution")
plt.xlabel("Value")
plt.ylabel("Frequency")

plt.show()
```

### Interpretation

The histogram forms a bell-shaped curve.

Most observations are concentrated near the mean (0), while fewer observations appear farther away.

This pattern is characteristic of a normal distribution and is frequently observed in natural phenomena.

Many machine learning techniques assume that numerical features approximately follow a normal distribution.

### Connection to Machine Learning

Normality is commonly checked during Exploratory Data Analysis (EDA).

Algorithms such as Linear Regression, Logistic Regression, and many statistical tests often perform better when numerical features are approximately normally distributed.

# Binomial Distribution

The binomial distribution models the number of successful outcomes in a fixed number of independent trials.

Each trial has only two possible outcomes:

- Success
- Failure

Examples include:

- Number of heads in 10 coin flips.
- Number of students who pass an exam.
- Number of successful emails delivered.

```python
# Simulate flipping a coin 10 times
binomial_data = np.random.binomial(
    n=10,
    p=0.5,
    size=10000
)
```

```python
plt.figure(figsize=(8,5))

plt.hist(binomial_data,
         bins=11,
         edgecolor="black")

plt.title("Binomial Distribution")
plt.xlabel("Number of Heads")
plt.ylabel("Frequency")

plt.show()
```

### Interpretation

Most experiments resulted in approximately five heads, which is expected because the probability of heads is 0.5.

The distribution is centered around the expected number of successes.

### Connection to Machine Learning

The binomial distribution is useful when modeling binary outcomes such as:

- Pass / Fail
- Spam / Not Spam
- Fraud / Not Fraud
- Diabetes / No Diabetes

# Uniform Distribution

A uniform distribution occurs when every possible outcome has the same probability.

Unlike the normal distribution, there is no central peak.

Examples include:

- Rolling a fair die.
- Random number generation.
- Random sampling.

```python
uniform_data = np.random.uniform(
    low=0,
    high=10,
    size=10000
)
```

```python
plt.figure(figsize=(8,5))

plt.hist(uniform_data,
         bins=20,
         edgecolor="black")

plt.title("Uniform Distribution")
plt.xlabel("Value")
plt.ylabel("Frequency")

plt.show()
```

### Interpretation

The histogram appears relatively flat because every value within the interval has approximately the same probability of occurring.

This is the defining characteristic of a uniform distribution.

# Comparing Common Probability Distributions

| Distribution | Shape | Common Example |
|--------------|-------|----------------|
| Normal | Bell-shaped and symmetric | Heights, exam scores |
| Binomial | Discrete distribution | Coin flips |
| Uniform | Flat distribution | Fair die roll, random numbers |

Each distribution describes a different type of random process, and selecting the appropriate distribution depends on the nature of the data being analyzed.

## Key Takeaways

- The **Normal Distribution** is continuous, symmetric, and commonly found in real-world measurements.
- The **Binomial Distribution** models the number of successes in repeated independent trials.
- The **Uniform Distribution** assigns equal probability to every possible outcome.

Recognizing these distributions helps data analysts choose appropriate statistical methods and better understand the characteristics of a dataset before building machine learning models.

# 🧪 Hands-On Lab: Probability & Distributions

## Objective

In this hands-on lab, we will apply probability concepts through simulations using NumPy.

Instead of relying only on mathematical formulas, we will generate random experiments and compare the simulated results with the theoretical expectations.

The lab demonstrates:

- The Law of Large Numbers.
- The behavior of the Normal Distribution.
- Conditional Probability using both manual calculation and simulation.

## Step 1: Simulating a Fair Coin

A fair coin has two equally likely outcomes:

- Heads
- Tails

The theoretical probability of getting Heads is:

$$
P(\text{Heads})=0.5
$$

We will simulate **10,000** coin flips and compare the observed proportion with the theoretical probability.

```python
coin_flips = np.random.choice(
    ["Heads", "Tails"],
    size=10000
)

heads = np.sum(coin_flips == "Heads")
tails = np.sum(coin_flips == "Tails")

print("Heads:", heads)
print("Tails:", tails)

print(f"Proportion of Heads: {heads/10000:.4f}")
```

```python
labels = ["Heads", "Tails"]
counts = [heads, tails]

plt.figure(figsize=(6,4))

plt.bar(labels, counts)

plt.title("Coin Flip Simulation (10,000 Flips)")
plt.ylabel("Count")

plt.show()
```

### Interpretation

The simulated proportion of heads should be close to **0.50**.

Although individual experiments are random, increasing the number of trials causes the observed probability to approach the theoretical probability.

This behavior illustrates the **Law of Large Numbers**, which states that experimental results become more accurate as the number of observations increases.

## Step 2: Simulating a Normal Distribution

We will generate random values from a normal distribution with:

- Mean = 0
- Standard Deviation = 1

The resulting histogram should display the familiar bell-shaped curve.

```python
normal_samples = np.random.normal(
    loc=0,
    scale=1,
    size=10000
)
```

```python
plt.figure(figsize=(8,5))

plt.hist(
    normal_samples,
    bins=30,
    edgecolor="black"
)

plt.title("Simulated Normal Distribution")
plt.xlabel("Value")
plt.ylabel("Frequency")

plt.show()
```

### Interpretation

The histogram forms a bell-shaped distribution centered around zero.

Most observations are concentrated near the mean, while fewer values appear farther from the center.

This pattern is characteristic of the normal distribution and is commonly observed in many natural and measurement-related phenomena.

## Step 3: Conditional Probability

Suppose a university has:

- 100 students.
- 40 students study Artificial Intelligence.
- 25 students study Artificial Intelligence and know Python.

We want to calculate:

$$
P(\text{Python}|\text{AI})
$$

using the conditional probability formula.

```python
students_ai = 40
students_python = 25

manual_probability = students_python / students_ai

print(f"Conditional Probability = {manual_probability:.3f}")
```

### Simulation

Next, we verify the same probability through a simple simulation.

```python
students = np.array(
    ["Python"] * 25 +
    ["No Python"] * 15
)

simulated_probability = np.mean(students == "Python")

print(f"Simulated Probability = {simulated_probability:.3f}")
```

### Interpretation

Both the manual calculation and the simulation produce approximately the same probability.

This confirms that simulation can effectively validate theoretical probability calculations.

In practice, simulations are useful when analytical solutions become difficult or impossible.
