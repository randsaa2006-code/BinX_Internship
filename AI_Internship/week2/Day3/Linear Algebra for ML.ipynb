# Matrices

A matrix is a two-dimensional arrangement of numbers organized into rows and columns.

In Machine Learning, a matrix is commonly used to represent an entire dataset.

Each **row** represents one data sample, while each **column** represents one feature.

This structure allows machine learning algorithms to process multiple samples efficiently using matrix operations.

This dataset can be represented as the following matrix:

\
\begin{bmatrix}

Rows represent customers (samples), while columns represent features.

# The Dot Product

The dot product is one of the most fundamental operations in linear algebra and machine learning.

It combines two vectors by multiplying their corresponding elements and then summing the results.

The final output is a single numerical value.

In machine learning, the dot product is used to combine feature values with model weights in order to generate predictions.

## Dot Product Formula

The dot product is calculated by multiplying each feature by its corresponding weight and then summing all the products.

$$
Prediction =
(25\times0.1)
+
(50000\times0.0002)
+
(3\times1.5)
$$

````python
features = np.array([25, 50000, 3])

weights = np.array([0.1, 0.0002, 1.5])

manual_prediction = (
    features[0] * weights[0] +
    features[1] * weights[1] +
    features[2] * weights[2]
)

print(manual_prediction)
````

````python
prediction = np.dot(features, weights)

print(prediction)
````

### Interpretation

Both the manual calculation and NumPy produced the same prediction.

The dot product combines feature values with their corresponding weights to produce a single output.

This output represents the prediction of a linear model before adding the bias term.

## Why Is the Dot Product Important?

The dot product is at the core of many machine learning algorithms.

For example:

- Linear Regression
- Logistic Regression
- Neural Networks

All of these models compute predictions by multiplying input features by learned weights.

## Prediction Formula

A linear model computes predictions using the following equation:

$$
Prediction = \mathbf{x}\cdot\mathbf{w} + b
$$

Where:
- $\mathbf{x}$ is the feature vector.
- $\mathbf{w}$ is the weight vector.
- $b$ is the bias term.

## Connection to Machine Learning

Every time a linear regression model predicts a value, it computes the dot product between the input features and the learned weights, then adds a bias term.

This same concept is also used in logistic regression and neural networks.

Understanding the dot product helps explain how machine learning models transform input features into predictions.

## Key Takeaways

- The dot product combines two vectors into a single value.
- Each feature is multiplied by its corresponding weight.
- The products are summed to produce the prediction.
- NumPy provides the `np.dot()` function to compute the dot product efficiently.
- The dot product is one of the most important mathematical operations in machine learning.

# Matrix Multiplication

Matrix multiplication is one of the most important operations in Machine Learning.

Instead of computing predictions for one sample at a time, matrix multiplication allows us to generate predictions for an entire dataset in a single operation.

This makes machine learning models much faster and more efficient when working with large datasets.

````python
features = np.array([25, 50000, 3])

weights = np.array([0.1, 0.0002, 1.5])

prediction = np.dot(features, weights)
print(prediction)
````

````python
X = np.array([
    [25,50000,3],
    [40,80000,10],
    [33,62000,5]
])

weights = np.array([
    0.1,
    0.0002,
    1.5
])
````

````python
print(X.shape)
print(weights.shape)
````

The dataset contains:

- 3 samples
- 3 features

The weight vector contains one weight for each feature.

Since the number of features equals the number of weights, the multiplication can be performed successfully.

````python
predictions = X @ weights

print(predictions)
````

### Interpretation

The matrix multiplication produced one prediction for each customer.

Instead of processing each sample individually, NumPy applied the dot product to every row in the dataset automatically.

This is exactly how machine learning models generate predictions for many samples efficiently.

## Shape Matching Rule

For matrix multiplication to work, the inner dimensions must match.

$$
(m \times n)
\times
(n \times p)
=
(m \times p)
$$

The number of columns in the first matrix must equal the number of rows in the second matrix.

### Valid Multiplication

$$
(3\times3)
\times
(3\times1)
=
(3\times1)
$$

✓ Multiplication is valid.

### Invalid Multiplication

$$
(3\times3)
\times
(2\times1)
$$

✗ Multiplication is not possible because

$$
3 \neq 2
$$

````python
wrong_weights = np.array([0.1,0.2])

X @ wrong_weights
````

### Why Did the Error Occur?

The dataset contains three features for every sample.

However, the weight vector contains only two values.

Because each feature must have a corresponding weight, the multiplication cannot be performed.

To fix the problem, the weight vector must contain exactly three values.

## Connection to Machine Learning

During prediction, a machine learning model multiplies the feature matrix by the learned weight vector.

This operation produces predictions for all samples simultaneously, making matrix multiplication one of the most frequently used operations in machine learning libraries such as NumPy, TensorFlow, and PyTorch.

## Key Takeaways

- Matrix multiplication computes predictions for multiple samples at once.
- The inner dimensions of the matrices must match.
- Shape mismatches are one of the most common programming errors in machine learning.
- Understanding matrix shapes helps prevent errors and makes it easier to work with machine learning models.

## Objective

In this lab, we will build a simple machine learning workflow using NumPy.

We will:

1. Represent a small dataset as a matrix.

4. Verify the calculations using NumPy.

This exercise demonstrates the mathematical operations performed internally by many machine learning models.

````python
import numpy as np

X = np.array([
    [25, 50000, 3],
    [40, 80000, 10],
    [33, 62000, 5]
])

print(X)
````

````python
print("Dataset Shape:", X.shape)
````

### Interpretation

The dataset contains:

- **3 samples** (rows)
- **3 features** (columns)

Each row represents one customer, while each column represents one feature.

## Step 2 — Create the Weight Vector

Assume that a machine learning model has learned the following weights:

- Age → 0.1
- Income → 0.0002
- Years of Employment → 1.5

These values represent the importance of each feature when generating predictions.

````python
weights = np.array([0.1, 0.0002, 1.5])

print(weights)
````

````python
print("Weight Shape:", weights.shape)
````

### Interpretation

The weight vector contains one weight for each feature.

Since our dataset has three features, the weight vector also contains three values.

## Step 3 — Manual Prediction

Before using NumPy, we will manually calculate the prediction for the first customer.

Customer Features:

- Age = 25
- Income = 50000
- Years = 3

````python
manual_prediction = (
    25 * 0.1 +
    50000 * 0.0002 +
    3 * 1.5
)

print(manual_prediction)
````

### Manual Calculation

$$
Prediction =
(25 \times 0.1)
+
(50000 \times 0.0002)
+
(3 \times 1.5)
$$

$$
=
2.5 + 10 + 4.5
=
17
$$

## Step 4 — Verify the Result with NumPy

Instead of calculating each multiplication manually, NumPy provides the `np.dot()` function.

````python
prediction = np.dot(X[0], weights)

print(prediction)
````

### Interpretation

The prediction produced by `np.dot()` is identical to the manual calculation.

This confirms that the dot product multiplies each feature by its corresponding weight and sums the results correctly.

## Step 5 — Matrix Multiplication

Rather than predicting one customer at a time, we can generate predictions for the entire dataset using matrix multiplication.

````python
predictions = X @ weights

print(predictions)
````

````python
for i, prediction in enumerate(predictions, start=1):
    print(f"Customer {i}: {prediction:.2f}")
````

### Interpretation

Matrix multiplication generated one prediction for every customer simultaneously.

Instead of calling `np.dot()` for each individual sample, NumPy automatically applied the dot product to every row in the dataset.

This approach is significantly more efficient and is commonly used in machine learning libraries.

## Step 6 — Shape Mismatch

A common programming mistake in machine learning is attempting to multiply matrices with incompatible shapes.

Let's intentionally create this error.

````python
wrong_weights = np.array([0.1, 0.2])

X @ wrong_weights
````

### Why Did This Error Occur?

The dataset contains **three features** for each sample.

However, the weight vector contains only **two values**.

Since every feature must have a corresponding weight, matrix multiplication cannot be performed.

To fix this problem, the weight vector must contain exactly three values.

````python
correct_weights = np.array([
    0.1,
    0.0002,
    1.5
])

predictions = X @ correct_weights

print(predictions)
````

### Interpretation

After providing one weight for each feature, the matrix multiplication succeeds.

Matching matrix dimensions is essential when working with machine learning models.
