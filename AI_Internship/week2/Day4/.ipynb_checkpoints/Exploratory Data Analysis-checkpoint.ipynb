# What is Exploratory Data Analysis (EDA)?

Exploratory Data Analysis (EDA) is the process of examining and understanding a dataset before applying machine learning algorithms.

Rather than building a model immediately, analysts first investigate the data to understand its characteristics, identify potential problems, and determine whether preprocessing is required.

EDA combines descriptive statistics with visualizations to answer important questions about the data.

Typical questions include:

- What does the dataset look like?
- How are the values distributed?
- Are there any missing values?
- Are there unusual or extreme observations?
- Are the variables balanced?

## Why is EDA Important?

Machine learning models learn patterns directly from the data they receive.

If the dataset contains errors, missing values, duplicated records, or unusual observations, the model may learn incorrect patterns and produce unreliable predictions.

Performing EDA before modeling helps identify these issues early and improves the overall quality of the analysis.

### Real-World Example

Imagine building a model to predict house prices.

Without performing EDA, you might overlook important issues such as:

- Houses with impossible prices.
- Missing information.
- Extremely large houses that behave as outliers.
- Features that contain little useful information.

Detecting these issues before training helps produce a more reliable machine learning model.

### Connection to Machine Learning

EDA is considered the first practical step in every machine learning project.

The insights gained during EDA help determine:

- Whether the data requires cleaning.
- Which features are informative.
- Whether outliers should be investigated.
- Whether transformations or feature engineering are needed before model training.

Well-performed EDA often leads to better model performance and more trustworthy predictions.

## Key Takeaways

- EDA is the process of understanding data before modeling.
- It helps identify patterns, distributions, and data quality issues.
- EDA combines statistical summaries with visualizations.
- Performing EDA before training a model improves the reliability of machine learning results.

# Univariate Analysis

Univariate analysis is the process of analyzing one variable at a time.

The goal is to understand the characteristics of a single feature without considering its relationship with other variables.

Through univariate analysis, we can identify:

- The distribution of values.
- The center of the data.
- The spread of observations.
- Potential outliers.
- Class frequencies for categorical variables.

This is usually the first analytical step in Exploratory Data Analysis (EDA).

## Why is Univariate Analysis Important?

Before studying relationships between variables, it is important to understand each feature individually.

Univariate analysis helps answer questions such as:

- Is the data normally distributed?
- Is the feature skewed?
- Are there any unusual observations?
- Are categorical classes balanced?
- Does the feature require preprocessing before modeling?

Understanding these characteristics leads to better feature engineering and more reliable machine learning models.

## Common Univariate Plots

| Plot | Purpose |
|------|---------|
| Histogram | Shows the distribution of numerical values |
| KDE Plot | Displays a smoothed estimate of the distribution |
| Box Plot | Detects spread and potential outliers |
| Count Plot | Shows the frequency of categorical values |

Suppose we have a dataset containing an **Age** column.

A histogram helps us understand how age values are distributed across the dataset.

```python
df = pd.read_csv("diabetes.csv")

df
```

```python
sns.histplot(data=df, x="Age", bins=20)

plt.title("Distribution of Age")
plt.xlabel("Age")
plt.ylabel("Frequency")

plt.show()
```

### Interpretation

A histogram can reveal:

- The most common values.
- Whether the data is symmetric.
- Whether the distribution is skewed.
- The presence of multiple peaks.
- Possible unusual values.

Histograms are often the first visualization created during EDA.

### Machine Learning Connection

Many machine learning algorithms assume that numerical variables follow approximately normal distributions.

Histograms help determine whether transformations such as logarithmic scaling or normalization may be necessary before training.

## KDE Plot

A Kernel Density Estimation (KDE) plot is a smoothed version of a histogram.

Instead of displaying bars, it estimates the probability density of the data using a smooth continuous curve.

KDE plots make it easier to identify the overall shape of a distribution.

```python
sns.kdeplot(data=df, x="Age", fill=True)

plt.title("Age Density")
plt.xlabel("Age")

plt.show()
```

### Interpretation

A KDE plot helps identify:

- Peaks in the data.
- Skewness.
- Multiple modes.
- Overall distribution shape.

Unlike histograms, KDE plots provide a smoother visualization that is often easier to interpret.

### Machine Learning Connection

KDE plots help analysts assess whether numerical features approximately follow a normal distribution, which is useful for selecting preprocessing techniques and statistical models.

## Box Plot

A box plot summarizes the distribution of a numerical variable using five key statistics:

- Minimum
- First Quartile (Q1)
- Median
- Third Quartile (Q3)
- Maximum

It also highlights potential outliers, making it one of the most useful tools for detecting unusual observations.

```python
sns.boxplot(data=df, x="Age")

plt.title("Box Plot of Age")

plt.show()
```

### Components of a Box Plot

A box plot contains:

- The line inside the box → Median
- The bottom of the box → First Quartile (Q1)
- The top of the box → Third Quartile (Q3)
- The whiskers → Typical range of the data
- Individual points outside the whiskers → Potential outliers

### Interpretation

A box plot quickly shows:

- The center of the data.
- The spread of observations.
- Whether the distribution is symmetric.
- Whether potential outliers exist.

It is commonly used before training machine learning models to investigate unusual values.

## Count Plot

A count plot is used for categorical variables.

It displays the number of observations belonging to each category.

```python
sns.countplot(data=df, x="Outcome")

plt.title("Outcome distribution")

plt.show()
```

### Interpretation

Count plots help identify:

- The most common categories.
- Rare categories.
- Class imbalance.

They provide a quick overview of the distribution of categorical variables.

### Machine Learning Connection

Many classification problems suffer from class imbalance.

Count plots allow analysts to detect imbalance early and decide whether techniques such as resampling or class weighting may be required.

## Key Takeaways

- Histograms show the distribution of numerical variables.
- KDE plots provide a smooth estimate of the distribution.
- Box plots summarize numerical data and highlight potential outliers.
- Count plots display the frequency of categorical variables.
- Univariate analysis is an essential first step before exploring relationships between variables or building machine learning models.

# Outlier Detection with the IQR Method

Outliers are observations that are significantly different from the rest of the data.

These unusual values may represent:

- Data entry errors.
- Measurement errors.
- Rare but valid observations.

Because outliers can strongly influence statistical analysis and machine learning models, they should always be investigated before making any decision.

## Why Are Outliers Important?

Outliers can affect many aspects of data analysis.

For example, they may:

- Increase the mean significantly.
- Increase the standard deviation.
- Distort the distribution of the data.
- Reduce the performance of some machine learning models.

However, not every outlier is incorrect.

Some outliers represent important real-world events and should be retained after careful investigation.

## The Interquartile Range (IQR)

The Interquartile Range (IQR) measures the spread of the middle 50% of the data.

It is calculated as:

$$
IQR = Q3 - Q1
$$

where:

- **Q1** is the first quartile (25th percentile).
- **Q3** is the third quartile (75th percentile).

### IQR Rule

A value is considered a potential outlier if it satisfies one of the following conditions:

$$
Value < Q1 - 1.5 \times IQR
$$

or

$$
Value > Q3 + 1.5 \times IQR
$$

Values outside these boundaries are flagged as potential outliers.

```python
Q1 = df["Insulin"].quantile(0.25)
Q3 = df["Insulin"].quantile(0.75)

IQR = Q3 - Q1

lower_bound = Q1 - 1.5 * IQR
upper_bound = Q3 + 1.5 * IQR

print("Lower Bound:", lower_bound)
print("Upper Bound:", upper_bound)
```

```python
outliers = df[
    (df["Insulin"] < lower_bound) |
    (df["Insulin"] > upper_bound)
]
```

```python
outliers
```

### Interpretation

The resulting DataFrame contains all observations that fall outside the acceptable IQR range.

These observations should be reviewed carefully before deciding whether they represent genuine values or data quality issues.

```python
sns.boxplot(data=df, x="Insulin")

plt.title("Insulin Box Plot")

plt.show()
```

### Interpretation

In a box plot:

- The box represents the middle 50% of the data.
- The line inside the box represents the median.
- The whiskers show the typical range of observations.
- Individual points outside the whiskers are potential outliers.

This visualization provides a quick overview of the distribution and highlights unusual observations.

## Should We Remove Outliers?

Outliers should never be removed automatically.

Before making any decision, it is important to investigate why they exist.

Possible actions include:

- Keep them if they represent valid observations.
- Correct them if they result from data entry errors.
- Remove them if they are clearly invalid.
- Cap or transform them if appropriate for the analysis.

Every decision should be documented and justified.

## Connection to Machine Learning

Outliers can significantly influence many machine learning algorithms.

For example:

- Linear Regression is highly sensitive to extreme values.
- Distance-based algorithms such as K-Nearest Neighbors (KNN) can also be affected.
- Tree-based models are generally more robust to outliers.

Identifying outliers during EDA helps analysts choose appropriate preprocessing techniques and build more reliable models.

## Key Takeaways

- Outliers are observations that differ significantly from the rest of the data.
- The IQR method is a simple and widely used technique for detecting potential outliers.
- Box plots provide a visual representation of outliers.
- Outliers should always be investigated before deciding whether to keep, modify, or remove them.
- Proper handling of outliers improves data quality and model reliability.

## Step 1 — Load the Dataset

## Step 2 — Histogram Analysis

Histograms help us understand how numerical values are distributed.

```python
numeric_columns = df.select_dtypes(include="number").columns

numeric_columns
```

```python
for column in numeric_columns:

    plt.figure(figsize=(6,4))

    sns.histplot(df[column], kde=True)

    plt.title(f"Distribution of {column}")

    plt.show()
```

### Interpretation

For each histogram, observe:

- Is the distribution approximately normal?
- Is it skewed?
- Are there multiple peaks?
- Are extreme values visible?

These observations help determine whether preprocessing or transformation may be necessary.

## Step 3 — KDE Analysis

Kernel Density Estimation (KDE) provides a smooth estimate of the data distribution.

```python
for column in numeric_columns:

    plt.figure(figsize=(6,4))

    sns.kdeplot(df[column], fill=True)

    plt.title(f"KDE Plot of {column}")

    plt.show()
```

### Interpretation

KDE plots make it easier to identify:

- Distribution shape.
- Peaks.
- Skewness.
- Possible multimodal distributions.

Compared to histograms, KDE plots provide a smoother representation of the data.

## Step 4 — Box Plot Analysis

Box plots summarize numerical variables and highlight potential outliers.

```python
for column in numeric_columns:

    plt.figure(figsize=(6,2))

    sns.boxplot(x=df[column])

    plt.title(f"Box Plot of {column}")

    plt.show()
```

### Interpretation

For each box plot, identify:

- The median.
- The spread of the data.
- Potential outliers.
- Whether the distribution appears symmetric.

Box plots are one of the fastest ways to detect unusual observations.

## Step 5 — Detect Outliers Using the IQR Method

We will detect potential outliers for one numerical feature using the IQR rule.

```python
column = "Insulin"
```

```python
Q1 = df[column].quantile(.25)

Q3 = df[column].quantile(.75)

IQR = Q3 - Q1
```

```python
lower = Q1 - 1.5 * IQR

upper = Q3 + 1.5 * IQR
```

```python
outliers = df[
    (df[column] < lower) |
    (df[column] > upper)
]

outliers
```

```python
print("Number of Outliers:", len(outliers))
```

### Interpretation

The IQR method identifies observations that fall outside the expected range.

These values should be investigated before deciding whether to keep, transform, or remove them.

Outliers are not necessarily incorrect—they may represent rare but meaningful observations.

## Step 6 — Count Plot Analysis

Count plots summarize the frequency of categorical variables.

```python
categorical_columns = ["Outcome"]

categorical_columns
```

```python
for column in categorical_columns:

    plt.figure(figsize=(6,4))

    sns.countplot(data=df, x=column)

    plt.title(f"Count Plot of {column}")

    plt.show()
```

### Interpretation

Count plots reveal:

- The most common categories.
- Rare categories.
- Possible class imbalance.

Class imbalance is especially important when building classification models.
